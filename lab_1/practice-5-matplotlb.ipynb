```python
import matplotlib.pyplot as plt
import numpy as np
# Simple line plot
x =np.linspace(0, 10, 100)
y =np.sin(x)
plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.title('Simple Sine Wave')
plt.xlabel('X values')
plt.ylabel('sin(x)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Multiple lines with different styles
x =np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)
plt.figure(figsize=(10, 6))
plt.plot(x, y1, 'r-', label='sin(x)', linewidth=2)
plt.plot(x, y2, 'b--', label='cos(x)', linewidth=2)
plt.plot(x, y3, 'g:', label='sin(x)*cos(x)', linewidth=2)
plt.title('Trigonometric Functions')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Bar chart
# Generate random data
np.random.seed(42)
x =np.random.randn(100)
y =2*x+np.random.randn(100) * 0.5
colors = np.random.rand(100)
sizes = np.random.rand(100) * 100
plt.figure(figsize=(8, 6))
plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap='viridis')
plt.title('Scatter Plot with Color and Size Variation')
plt.xlabel('X')
plt.ylabel('Y')
plt.colorbar(label='Color Value')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Box plot(Histogram)
# Generate normal distribution data
data = np.random.randn(1000)
plt.figure(figsize=(10, 6))
plt.hist(data, bins=30, alpha=0.7, color='steelblue', edgecolor='black')
plt.title('Histogram of Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Box plot(Box and Whisker Plot)
# Create multiple plots in one figure
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
# Top-left: Line plot
x =np.linspace(0, 10, 100)
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title('Sine')
axes[0, 0].grid(True)
# Top-right: Scatter plot
np.random.seed(42)
axes[0, 1].scatter(np.random.randn(50), np.random.randn(50), alpha=0.6)
axes[0, 1].set_title('Random Scatter')
axes[0, 1].grid(True)
# Bottom-left: Histogram
data = np.random.randn(1000)
axes[1, 0].hist(data, bins=30, alpha=0.7)
axes[1, 0].set_title('Histogram')
axes[1, 0].grid(True)
# Bottom-right: Bar chart
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]
axes[1, 1].bar(categories, values, color='green', alpha=0.7)
axes[1, 1].set_title('Bar Chart')
axes[1, 1].grid(True, axis='y')
plt.tight_layout()
plt.show()
```
