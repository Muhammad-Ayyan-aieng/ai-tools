```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
print("All imports successful!")
```

```python
x=np.linspace(0,10,100)
y=np.sin(x)
plt.figure(figsize=(8, 4))
plt.plot(x, y, 'b-', linewidth=2)
plt.title('My First Plot in Jupyter')
plt.xlabel('X values')
plt.ylabel('sin(x)')
plt.grid(True, alpha=0.3)
plt.show()
```
