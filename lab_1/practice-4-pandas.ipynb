```python
# Create from dictionary
data = {
'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
'Age': [25, 30, 35, 28],
'Salary': [50000, 60000, 70000, 55000],
'Department': ['HR', 'IT', 'IT', 'Finance']
}
df = pd.DataFrame(data)
print(df)
```

```python
# Handling missing data
# Create a messy dataset
messy_data = {
'Name': ['Alice', 'Bob', None, 'Diana', 'Eve'],
'Age': [25, None, 35, 28, None],
'Salary': [50000, 60000, None, 55000, 65000],
'Join_Date': ['2020-01-15', '2019-03-20', '2021-06-01', None, '2022-11-10']
}
df_messy = pd.DataFrame(messy_data)
print(df_messy)
# Check for missing values
print(df_messy.isnull())
print(df_messy.isnull().sum())
# Drop rows with any missing values
df_clean_drop = df_messy.dropna()
print(df_clean_drop)
# Drop rows where specific column is missing
df_clean_specific = df_messy.dropna(subset=['Name', 'Salary'])
print(df_clean_specific)
# Fill missing values
df_filled = df_messy.copy()
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())
df_filled['Name'] = df_filled['Name'].fillna('Unknown')
print(df_filled)
# Fill with different strategies
df_filled['Salary'] = df_filled['Salary'].fillna(method='ffill') # Forward fill
df_filled['Join_Date'] = df_filled['Join_Date'].fillna('2023-01-01') # Specific value
```

```python
# Data manipulation and transformation
# Add new columns
df['Bonus'] = df['Salary'] * 0.1
df['Seniority'] = df['Age'].apply(lambda age: 'Senior' if age >= 30 else 'Junior')
# Apply function to column
df['Name_Upper'] = df['Name'].str.upper()
# Rename columns
df_renamed = df.rename(columns={'Name': 'Employee Name', 'Salary': 'Annual Salary'})
# Create categorical
df['Department'] = pd.Categorical(df['Department'])
# One-hot encoding (for machine learning)
dummies = pd.get_dummies(df['Department'], prefix='Dept')
df_with_dummies = pd.concat([df, dummies], axis=1)
# Group by operations
dept_stats = df.groupby('Department').agg({
'Salary': ['mean', 'min', 'max', 'count'],
'Age': 'mean'
})
print(dept_stats)
# Pivot tables
pivot = pd.pivot_table(df,
values='Salary',
index='Department',
columns='Seniority',
aggfunc='mean')
print(pivot)
```

```python
# Merging and joining DataFrames
# Create two DataFrames
df1 = pd.DataFrame({
'ID': [1, 2, 3],
'Name': ['Alice', 'Bob', 'Charlie']
})
df2 = pd.DataFrame({
'ID': [2, 3, 4],
'Score': [85, 92, 78]
})
# Inner join (only matching IDs)
inner = pd.merge(df1, df2, on='ID', how='inner')
print(inner) # IDs 2 and 3
# Left join (keep all from df1)
left = pd.merge(df1, df2, on='ID', how='left')
print(left) # ID 1 has NaN for Score
# Right join (keep all from df2)
right = pd.merge(df1, df2, on='ID', how='right')
print(right) # ID 4 has NaN for Name
# Outer join (keep all from both)
outer = pd.merge(df1, df2, on='ID', how='outer')
print(outer) # All IDs, NaN where missing
```
