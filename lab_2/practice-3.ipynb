```python
import numpy as np
import pandas as pd
# Create a realistic dataset with missing values
np.random.seed(42)
n_samples = 1000
# Generate data with different missing patterns
data = {
'customer_id': range(1000, 1000 + n_samples),
'age': np.random.normal(40, 15, n_samples),
'income': np.random.normal(50000, 20000, n_samples),
'credit_score': np.random.normal(700, 50, n_samples),
'purchases': np.random.poisson(5, n_samples),
'satisfaction': np.random.uniform(1, 5, n_samples),
'region': np.random.choice(['North', 'South', 'East', 'West'], n_samples),
'member_since': pd.date_range('2010-01-01', periods=n_samples, freq='D')
}
df = pd.DataFrame(data)
# Introduce missing values in different patterns
# MCAR(Missing Completely At Random)
mcar_idx = np.random.choice(n_samples, 50, replace=False)
df.loc[mcar_idx, 'age'] = np.nan
# MAR(Missing At Random)- missing depends on observed data
# Older customers more likely to have missing income
mar_idx = df[df['age'] > 50].sample(30).index
df.loc[mar_idx, 'income'] = np.nan
# MNAR(Missing Not At Random)- missing depends on unobserved value
# Customers with low satisfaction less likely to report it
mnar_idx = df[df['satisfaction'] < 2].sample(20).index
df.loc[mnar_idx, 'satisfaction'] = np.nan
# Random missing in other columns
df.loc[np.random.choice(n_samples, 40, replace=False), 'credit_score'] = np.nan
df.loc[np.random.choice(n_samples, 30, replace=False), 'purchases'] = np.nan
print("Dataset with missing values created!")
print(df.isnull().sum())
```
